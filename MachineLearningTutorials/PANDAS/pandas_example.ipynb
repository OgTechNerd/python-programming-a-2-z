```python
import pandas as pd 
import numpy as np
import sys
print(sys.version)
print(pd.__version__)
```

Series: a one-dimensional labeled array holding data of any type such as integers, strings, Python objects etc.
DataFrame: a two-dimensional data structure that holds data like a two-dimension array or a table with rows and columns.
Creating a Series by passing a list of values, letting pandas create a default RangeIndex.

```python
s = pd.Series([1, 3, 5, np.nan, 6, 8])
print(s)
```

Creating a DataFrame by passing a NumPy array with a datetime index using date_range() and labeled columns:

```python
dates = pd.date_range("20200101", periods=4)
iter = ["column1", "column2", "coloumn3", "column4"]
df = pd.DataFrame(np.random.randn(4, 4), index=dates, columns=list([iter]))   
print(df)
```

Creating a DataFrame by passing a dictionary of objects where the keys are the column labels and the values are the column values.

```python
df1 = pd.DataFrame(
    {
        "A": 1.0,
        "B": pd.Timestamp("20130102"),
        "C": pd.Series(1, index=list(range(4)), dtype="float32"),
        "D": np.array([3] * 4, dtype="int32"),
        "E": pd.Categorical(["test", "train", "test", "train"]),
        "F": "foo",
    }
)
print(df1)
df1.dtypes
```
